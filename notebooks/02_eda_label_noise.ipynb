# EDA — Ca nghi ngờ nhiễu nhãn (Confident Learning)

Thể hiện kết quả phân tích **Confident Learning** (cleanlab) trên OOF predictions của mô hình
(cells 18/19, `notebook2d3557181f (8).ipynb`):

- **Pie chart**: tỷ lệ **48.1%** ca nghi ngờ nhiễu nhãn trong bộ ca được đánh giá.
- **Bar chart**: **65%** ca nhiễu tập trung ở **vùng ranh giới BI-RADS 1 vs 2–3**.

Các con số là kết quả phân tích đã chạy trên Kaggle (dữ liệu VinDr-Mammo), được giữ dưới dạng
hằng số để dễ chỉnh sửa — notebook này **không** nạp dữ liệu.


```python
# =============================================================================
# CELL 1 — IMPORTS & REPRODUCIBILITY
# =============================================================================
import numpy as np
import matplotlib.pyplot as plt

_GLOBAL_SEED = 42
np.random.seed(_GLOBAL_SEED)
```

```python
# =============================================================================
# CELL 2 — SỐ LIỆU TÓM TẮT (HẰNG SỐ, KHÔNG NẠP DỮ LIỆU)
# =============================================================================
# Nguồn: Confident Learning (cleanlab) trên OOF predictions, cells 18/19 của
# notebook2d3557181f (8).ipynb (chạy trên Kaggle). Chỉnh đúng số khi có kết quả
# chạy lại.

# --- Pie: tỷ lệ ca nghi ngờ nhiễu nhãn ---
NOISE_RATIO   = 48.1   # % ca nghi ngờ nhiễu nhãn
CLEAN_RATIO   = 100.0 - NOISE_RATIO   # 51.9 % còn lại

# --- Bar: vị trí tập trung của ca nhiễu ---
BOUNDARY_1_23 = 65.0   # % ca nhiễu ở ranh giới BI-RADS 1 vs 2-3
BOUNDARY_OTHERS = 100.0 - BOUNDARY_1_23   # 35.0 % ca nhiễu ở vùng khác

print(f"Pie   — Nghi ngờ nhiễu nhãn: {NOISE_RATIO:.1f}% | Nhãn sạch: {CLEAN_RATIO:.1f}%")
print(f"Bar   — Ranh giới BI-RADS 1 vs 2-3: {BOUNDARY_1_23:.1f}% | Vùng khác: {BOUNDARY_OTHERS:.1f}%")
```

## Hình 2 — Pie Chart (tỷ lệ ca nghi ngờ nhiễu nhãn) + Bar Chart (vị trí tập trung ca nhiễu)


```python
# =============================================================================
# CELL 3 — PIE CHART & BAR CHART (2 PANEL KỀ NHAU)
# =============================================================================
_C_NOISE  = "#D7263D"   # đỏ — nghi ngờ nhiễu nhãn
_C_CLEAN  = "#2E86AB"   # xanh dương — nhãn sạch
_C_BOUNDARY = "#F5B041"  # cam — ranh giới BI-RADS 1 vs 2-3
_C_OTHER   = "#A1B8C7"   # xám — vùng khác

fig, (ax_pie, ax_bar) = plt.subplots(1, 2, figsize=(16, 6.6))
fig.subplots_adjust(top=0.83, bottom=0.20, wspace=0.50)

# ---------- Trái: Pie Chart ----------
sizes = [NOISE_RATIO, CLEAN_RATIO]
labels_pie = ["Nghi ngờ nhiễu nhãn", "Nhãn sạch"]
colors_pie = [_C_NOISE, _C_CLEAN]

# Thu nhỏ pie để nhường chỗ cho legend ở phía dưới
wedges, texts, autotexts = ax_pie.pie(
    sizes, labels=None, colors=colors_pie, autopct="%.1f%%",
    startangle=140, counterclock=False,
    explode=(0.05, 0.0), radius=0.70,
    wedgeprops=dict(edgecolor="white", linewidth=2),
    textprops=dict(fontsize=11),
)
for t in autotexts:
    t.set_color("white"); t.set_fontweight("bold")
ax_pie.set_title("Ca nghi ngờ nhiễu nhãn", fontsize=12, fontweight="bold")

# Legend đặt dưới đáy pie (nằm gọn trong panel trái, không tràn sang panel phải)
leg = ax_pie.legend(wedges, [f"{l} — {s:.1f}%" for l, s in zip(labels_pie, sizes)],
                    loc="upper center", bbox_to_anchor=(0.5, -0.18),
                    frameon=True, fontsize=10, ncol=2)

# ---------- Phải: Bar Chart ----------
x = np.arange(2)
bar_labels = ["BI-RADS 1\nvs 2–3\n(ranh giới)", "Các vùng\nkhác"]
bar_values = [BOUNDARY_1_23, BOUNDARY_OTHERS]
bar_colors = [_C_BOUNDARY, _C_OTHER]

bars = ax_bar.bar(x, bar_values, width=0.55, color=bar_colors,
                  edgecolor="black", linewidth=0.7)
for xi, v in zip(x, bar_values):
    ax_bar.text(xi, v + 2, f"{v:.1f}%", ha="center", va="bottom",
                fontsize=14, fontweight="bold")

ax_bar.set_ylim(0, 100)
ax_bar.set_xticks(x)
ax_bar.set_xticklabels(bar_labels, fontsize=11)
ax_bar.set_ylabel("Tỷ lệ ca nhiễu nhãn (%)", fontsize=11)
ax_bar.set_title("Vị trí tập trung của ca nhiễu", fontsize=12, fontweight="bold")
ax_bar.grid(axis="y", alpha=0.3)
ax_bar.spines["top"].set_visible(False)
ax_bar.spines["right"].set_visible(False)

fig.suptitle("Hình 2 — 48.1% ca nghi ngờ nhiễu nhãn; 65% ca nhiễu tập trung ở ranh giới BI-RADS 1 vs 2–3",
             fontsize=13, fontweight="bold", y=0.97)
plt.show()
```
